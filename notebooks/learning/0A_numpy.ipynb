> **NumPy library**



```python
import numpy as np
```

> **Basics**

```python
# Vector
vector = np.array([1,2,3], dtype="int16") 
print(vector)
```

```python
# Matrix
matrix = np.array([[1,2,3],[4,5,6]])
print(matrix)
```

```python
# Shape
matrix.shape
```

```python
# Dimensions
matrix.ndim
```

```python
# Type
vector.dtype
```

```python
# Size (number of bytes)
vector.itemsize
```

```python
# total size n_bytes * elements
vector.nbytes
```

> **Accessing/changing elements, rows, columns, ...**

```python
print(matrix)
```

```python
# Specific element [r,c], r=rows, c=columns
matrix[0,0]
```

```python
matrix[1,2]
```

```python
# Row
matrix[0,:]
```

```python
# Column
matrix[:,1]
```

```python
# Secifics elements [startindex:endindex:stepsize]
matrix[0, 1:3]
```

```python
matrix[:, 0:2]
```

```python
matrix[:,-1]
```

```python
# Change value
matrix[0,0] = 10
print(matrix)
```

```python
# Change vectors
matrix[:, -1] = [100,100]
print(matrix)
```

> **Initializing arrays**

```python
# All zeros
np.zeros((2,2), dtype="int8")
```

```python
# All ones
np.ones((1,3,4))
```

```python
# Other number
np.full((3,3), 100)
```

```python
# Other number
np.full_like((3,3), 11)
```

```python
# Random numbers
np.random.rand(4,2)
```

```python
# Random ints

np.random.randint(5,10, size=(3,3))
```

```python
np.identity(5)
```

```python
arr = np.array([1,2,3])
np.repeat(arr,3,axis=0)
```

```python
arr = np.array([[[1,2,3]]])
np.repeat(arr,3,axis=2)
```

```python
example = np.ones((5,5),dtype="int8")
print(example)
```

```python
example[1:4,1:4] = 0
example[2,2] = 9
example
```

> **Copying arrays**

```python
a = np.array([1,2,3])
b = a
b
```

```python
b[0] = 100
print(b, a)
```

```python
a = np.array([1,2,3])
b = a.copy()
b
```

```python
b[0] = 100
print(b, a)
```
